Problem 2: Savings in 20 years under uncertain returns

If I save a fixed amount each month for 20 years and market returns are random, how much money could I end up with and how much must I save to be 90 percent sure of reaching 150,000 euros?

Variables: monthly saving in Euros

Objective: Minimize the monthly saving amount

Constraints:
1) The chance of ending with at least 150,000 euros after 20 years must be atleast 90 percent
2) The monthly amount can't be negative.

Assumptions: 
1) The values are calculated with the 6% return, 15% volatility and 10,000 simulated futures. 
2) The returns are random draws, not a forecast.
3) The monthly saving stays fixed for all 20 years, and the returns are assumed to follow a normal distribution.


```python
import numpy as np
```

```python
monthly_saving = 300
```

```python
years = 20
avg_return = 0.06
volatility = 0.15
```

```python
n_sims = 10000
```

```python
returns = np.random.normal(avg_return, volatility, (n_sims, years))
```

```python
print(returns.shape)
```

```python
balance = np.zeros(n_sims)
```

```python
for year in range(years):
    balance = balance * (1 + returns [: , year])
    balance = balance + 12 * monthly_saving
```

```python
print(np.percentile(balance, [10, 50, 90]))
```

```python
def final_balances(monthly):
    balance = np.zeros(n_sims)
    for year in range(years):
        balance = balance * (1 + returns[ :, year]) + 12 * monthly
    return balance
```

```python
result = final_balances(300)
print(np.percentile(result, 10))
```

```python
for monthly in range(300, 801, 50):
    print(monthly, np.percentile(final_balances(monthly), 10))
```

Observation: At 300 euros a month, the typical outcome was about 121,000 euros, but a 90 percent chance of reaching 150,000 euros needed about 650 euros a month. So, guaranteeing a target costs far more than aiming for the typical outcome, because you are protecting against bad market years.
Note: I searched in steps of 50 euros, so the true minimum lies between 600 and 650 euros a month. A finer search would narrow it down.
