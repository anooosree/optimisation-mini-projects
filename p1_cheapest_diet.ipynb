Problem 1: The cheapest balanced diet
I would like to eat enough protein and calories each day while spending as little as possible.

Variables (grams of each food eaten per day): lentils, rice, tofu, spinach, peanuts

Objective: Minimize the total daily cost

Constraints:
1) Total protein must be atleast a minimum of 60 grams per day.
2) Total calories must be atleast around 2000 per day.
3) No food can have a negative amount.

Assumptions: The prices and nutrition values are rough estimates per 100 grams.

All amounts are measured in units of 100 grams, so a value of 1 means 100 grams of that food. 

```python
import numpy as np
from scipy.optimize import linprog
```

```python
# Price in Euros per 100 grams
c = [0.20, 0.15, 0.40, 0.25, 0.30] 
```

```python
# Grams of protein per 100 grams
protein = [ 24, 7, 13, 3, 26]
```

```python
# Calories per 100 grams
calories = [350, 350, 120, 25, 600]
```

```python
# linprog only accepts atmost values, so the atleast values rules for protein and calories are multiplied by minus one
# A holds protein and calorie rows, b holds the limits minus 60 and minus 2000

A = [[-24, -7, -13, -3, -26], [-350, -350, -120, -25, -600]]
b = [-60, -2000]
```

```python
result = linprog(c, A_ub = A, b_ub = b, bounds = [(0, None)] * 5)
```

```python
print(result.x)
```

```python
print(result.fun)
```

Part 2: Requiring atleast 100 grams of Spinach

```python
result2 = linprog(c, A_ub = A, b_ub = b, bounds = [(0, None), (0, None), (0, None), (1, None), (0, None)])
```

```python
print(result2.x)
```

```python
print(result2.fun)
```

Observation: The price before adding the spinach requirement is 0.92(in Euros) and after is 1.15( in Euros). So, adding a spinach requirement increased the minimum daily cost by about 25 percent. 

The solver ignored the spinach before because it was not the cheapest source of protein or calories.
